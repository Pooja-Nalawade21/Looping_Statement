```python
#CHECH WHETHER NO. IS PRIME OR NOT
num=int(input("Enter a Number : "))
for i in range(2, num+1):
    count = 0
    for j in range(1, i + 1):
        if i % j == 0:
            count += 1
if count == 2:
    print("The Given Number is Prime.")
else :
    print("The Given Number is not  Prime.")
        
```
